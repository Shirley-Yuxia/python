**4.Gather daily confirmed cases worldwide. (Hint: summarise daily confirmed cases over all countries.)**

```python
cases_list = deaths_df.iloc[:, 4:] # gather the number of cases worldwide starting from 2nd row and 5th column
print(cases_list)#give the list to find out all the cases worldwide
```

```python
print('The confirmed cases for each day is')# A statement for next data
print(cases_list.sum())#list out the sum of daily confirmed cases worldwide

total_deaths = cases_list.sum().sum()#calculation the total of deaths in each day
print('The total number of deaths worldwide upto 03/09/2023 is:', total_deaths)
```

```python
row_values = cases_list.sum(axis=1)#give the list to find out all the cases each contries
print(row_values)

countries_names = deaths_df['Country/Region']# list out the name of countries
print(countries_names)

result= pd.concat([countries_names,row_values], axis=1)#combine the two sets of data into one list
print('The confirmed cases of each countries are ')
result
```
